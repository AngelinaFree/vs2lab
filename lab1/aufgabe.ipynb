```python
import os
import aufgabe1

server = aufgabe1.Server()
pid = os.fork()

if pid == 0:
    # Kindprozess: Server bedienen
    try:
        server.serve()
    finally:
        os._exit(0)
else:
    # Elternprozess
    print("Server läuft mit PID:", pid)
```
