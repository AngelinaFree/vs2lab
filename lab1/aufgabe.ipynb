## Aufsetzten des Servers

wir forken den Prozess, damit der Server weiterhin laufen kann, auf wenn wir später diesem auf dem Main-Thred Client Aufgaben geben

```python
import os
import aufgabe1

server = aufgabe1.Server()
pid = os.fork()

if pid == 0:
    try:
        server.serve()
    finally:
        os._exit(0)
```
